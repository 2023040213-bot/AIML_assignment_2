```python
# Import required libraries

import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score
```

```python
# Load wine quality dataset

df = pd.read_csv("winequalityN.csv")

df.head()
```

```python
# Fill missing numerical values

numeric_columns = df.select_dtypes(include=["int64", "float64"]).columns

df[numeric_columns] = df[numeric_columns].fillna(
    df[numeric_columns].median()
)

# Remove wine type because it is categorical
# and is not required for this classification

X = df.drop(["quality", "type"], axis=1)

# Target variable

y = df["quality"]

X.head()
```

```python
# Split data into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
# Create Naïve Bayes classifier

nb_model = GaussianNB()

# Train the model

nb_model.fit(X_train, y_train)

# Predict test data

nb_pred = nb_model.predict(X_test)
```

```python
# Create kNN classifier

knn_model = KNeighborsClassifier(n_neighbors=5)

# Train the model

knn_model.fit(X_train, y_train)

# Predict test data

knn_pred = knn_model.predict(X_test)
```

```python
# Calculate performance metrics for Naïve Bayes

nb_accuracy = accuracy_score(y_test, nb_pred)

nb_precision = precision_score(
    y_test,
    nb_pred,
    average="weighted",
    zero_division=0
)

nb_recall = recall_score(
    y_test,
    nb_pred,
    average="weighted",
    zero_division=0
)


# Calculate performance metrics for kNN

knn_accuracy = accuracy_score(y_test, knn_pred)

knn_precision = precision_score(
    y_test,
    knn_pred,
    average="weighted",
    zero_division=0
)

knn_recall = recall_score(
    y_test,
    knn_pred,
    average="weighted",
    zero_division=0
)
```

```python
# Display performance of both algorithms

results = pd.DataFrame({
    "Algorithm": ["Naïve Bayes", "kNN"],
    "Accuracy": [nb_accuracy, knn_accuracy],
    "Precision": [nb_precision, knn_precision],
    "Recall": [nb_recall, knn_recall]
})

results
```
