```python
# Import required libraries

import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score
```

```python
# Load customer churn dataset

df = pd.read_csv("customer_churn.csv")

df.head()
```

```python
# Remove customer ID if present

if "customerID" in df.columns:
    df = df.drop("customerID", axis=1)

# Convert categorical columns into numbers

encoder = LabelEncoder()

for column in df.select_dtypes(include="str").columns:
    df[column] = encoder.fit_transform(df[column])

# Separate input and target

X = df.drop("Churn", axis=1)
y = df["Churn"]

# Fill missing values

X = X.fillna(X.median())

X.head()
```

```python
# Split data into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
# Create and train Naïve Bayes classifier

nb_model = GaussianNB()

nb_model.fit(X_train, y_train)

# Predict test data

nb_pred = nb_model.predict(X_test)
```

```python
# Create and train kNN classifier

knn_model = KNeighborsClassifier(n_neighbors=5)

knn_model.fit(X_train, y_train)

# Predict test data

knn_pred = knn_model.predict(X_test)
```

```python
# Calculate metrics for Naïve Bayes

nb_accuracy = accuracy_score(y_test, nb_pred)

nb_precision = precision_score(
    y_test, nb_pred, zero_division=0
)

nb_recall = recall_score(
    y_test, nb_pred, zero_division=0
)


# Calculate metrics for kNN

knn_accuracy = accuracy_score(y_test, knn_pred)

knn_precision = precision_score(
    y_test, knn_pred, zero_division=0
)

knn_recall = recall_score(
    y_test, knn_pred, zero_division=0
)
```

```python
# Compare Naïve Bayes and kNN

results = pd.DataFrame({
    "Algorithm": ["Naïve Bayes", "kNN"],
    "Accuracy": [nb_accuracy, knn_accuracy],
    "Precision": [nb_precision, knn_precision],
    "Recall": [nb_recall, knn_recall]
})

results
```

```python
# Test first 5 records from the test dataset

test_data = X_test.iloc[:5]

comparison = pd.DataFrame({
    "Actual": y_test.iloc[:5].values,
    "Naive Bayes": nb_model.predict(test_data),
    "kNN": knn_model.predict(test_data)
})

comparison
```
