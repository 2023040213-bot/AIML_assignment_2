```python
# Import required libraries

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
```

```python
# Load the Heart Disease dataset

df = pd.read_csv("heart.csv")

# Display first 5 records

df.head()
```

```python
# Display all column names

df.columns
```

```python
# Display information about the dataset

df.info()
```

```python
# X contains independent/input variables
# y contains the dependent/output variable

X = df.drop("heartdisease", axis=1)

y = df["heartdisease"]

# Display input variables

X.head()
```

```python
# Split data into training and testing sets
# 80% training data
# 20% testing data

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Display shapes

X_train.shape, X_test.shape
```

```python
# ----------------------------------------------------------
# Bayesian Network Structure
# ----------------------------------------------------------

# The target node is HeartDisease.
# Medical features are connected to the target.

network = {
    "age": "HeartDisease",
    "sex": "HeartDisease",
    "cp": "HeartDisease",
    "trestbps": "HeartDisease",
    "chol": "HeartDisease",
    "fbs": "HeartDisease",
    "restecg": "HeartDisease",
    "thalach": "HeartDisease",
    "exang": "HeartDisease",
    "oldpeak": "HeartDisease",
    "slope": "HeartDisease",
    "ca": "HeartDisease",
    "thal": "HeartDisease"
}

network
```

```python
# ----------------------------------------------------------
# Create Conditional Probability Tables
# ----------------------------------------------------------

# Combine training input and target

train_data = X_train.copy()

train_data["target"] = y_train.values

# Store conditional probability tables

probability_tables = {}

# Features used in our Bayesian network

features = X.columns

# Calculate P(feature | target)

for feature in features:
    
    probability_tables[feature] = pd.crosstab(
        train_data[feature],
        train_data["target"],
        normalize="columns"
    )

probability_tables
```

```python
# Display conditional probability table for chest pain

probability_tables["cp"]
```

```python
# ----------------------------------------------------------
# Calculate prior probability
# ----------------------------------------------------------

prior_probability = y_train.value_counts(normalize=True)

prior_probability
```

```python
# ----------------------------------------------------------
# Bayesian prediction function
# ----------------------------------------------------------

def predict_bayesian(row):
    
    probabilities = {}
    
    # Calculate probability for each target class
    
    for target_value in [0, 1]:
        
        # Start with prior probability
        
        probability = prior_probability.get(target_value, 0)
        
        # Multiply conditional probabilities
        # P(feature | target)
        
        for feature in features:
            
            value = row[feature]
            
            table = probability_tables[feature]
            
            if value in table.index and target_value in table.columns:
                conditional_probability = table.loc[value, target_value]
            else:
                conditional_probability = 0.0001
            
            # Avoid zero probability
            conditional_probability = max(
                conditional_probability,
                0.0001
            )
            
            probability *= conditional_probability
        
        probabilities[target_value] = probability
    
    # Select class having maximum probability
    
    return max(
        probabilities,
        key=probabilities.get
    )
```

```python
# Predict one patient from test data

single_patient = X_test.iloc[0]

single_prediction = predict_bayesian(single_patient)

single_prediction
```

```python
# Predict all patients in the testing dataset

y_pred = X_test.apply(
    predict_bayesian,
    axis=1
)

y_pred.head(10)
```

```python
# Calculate accuracy

accuracy = accuracy_score(
    y_test,
    y_pred
)

accuracy
```

```python
# Accuracy in percentage

accuracy * 100
```

```python
# Calculate precision

precision = precision_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

precision
```

```python
# Calculate recall

recall = recall_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

recall
```

```python
# Create a result table

results = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Precision",
        "Recall"
    ],
    
    "Score": [
        accuracy,
        precision,
        recall
    ]
})

results
```

```python
# Display evaluation metrics as percentages

results_percentage = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Precision",
        "Recall"
    ],
    
    "Percentage": [
        accuracy * 100,
        precision * 100,
        recall * 100
    ]
})

results_percentage
```
